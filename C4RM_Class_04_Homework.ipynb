```python
def getBondPrice(y, face, couponRate, m, ppy=1):
    periods = int(m * ppy)
    coupon = face * couponRate / ppy
    bondPrice = 0.0

    for k in range(1, periods + 1):
        cashFlow = coupon
        if k == periods:
            cashFlow += face
        bondPrice += cashFlow / (1 + y / ppy) ** k

    return bondPrice


y = 0.03
face = 2000000
couponRate = 0.04
m = 10
print(f"Annual (ppy=1): {getBondPrice(y, face, couponRate, m, 1):,.6f}")
print(f"Semiannual (ppy=2): {getBondPrice(y, face, couponRate, m, 2):,.6f}")
print(f"Default ppy: {getBondPrice(y, face, couponRate, m):,.6f}")
```

```python
def getBondDuration(y, face, couponRate, m, ppy=1):
    periods = int(m * ppy)
    coupon = face * couponRate / ppy
    bondPrice = 0.0
    timeWeightedPV = 0.0

    for k in range(1, periods + 1):
        cashFlow = coupon
        if k == periods:
            cashFlow += face
        presentValue = cashFlow / (1 + y / ppy) ** k
        bondPrice += presentValue
        timeWeightedPV += (k / ppy) * presentValue

    bondDuration = timeWeightedPV / bondPrice
    return bondDuration


print(f"Annual duration: {getBondDuration(y, face, couponRate, m, 1):.9f} years")
print(f"Semiannual duration: {getBondDuration(y, face, couponRate, m, 2):.9f} years")
```

```python
def getBondPrice_E(face, couponRate, m, yc):
    if len(yc) != m:
        raise ValueError("yc must contain one rate for each year.")

    bondPrice = 0.0
    for t, rate in enumerate(yc, start=1):
        cashFlow = face * couponRate
        if t == m:
            cashFlow += face
        bondPrice += cashFlow / (1 + rate) ** t

    return bondPrice


yc = [0.010, 0.015, 0.020, 0.025, 0.030]
print(f"Enumerate price: {getBondPrice_E(2000000, 0.04, 5, yc):,.6f}")
```

```python
def getBondPrice_Z(face, couponRate, times, yc):
    if len(times) != len(yc) or len(times) == 0:
        raise ValueError("times and yc must have the same nonzero length.")

    bondPrice = 0.0
    for t, rate in zip(times, yc):
        bondPrice += face * couponRate / (1 + rate) ** t

    bondPrice += face / (1 + yc[-1]) ** times[-1]
    return bondPrice


times = [1, 1.5, 3, 4, 7]
print(f"Zip price: {getBondPrice_Z(2000000, 0.04, times, yc):,.6f}")
```

```python
def FizzBuzz(start, finish):
    outlist = []
    for i in range(start, finish + 1):
        if i % 15 == 0:
            outlist.append("fizzbuzz")
        elif i % 3 == 0:
            outlist.append("fizz")
        elif i % 5 == 0:
            outlist.append("buzz")
        else:
            outlist.append(i)
    return outlist


print(FizzBuzz(1, 15))
print(FizzBuzz(14, 16))
```
